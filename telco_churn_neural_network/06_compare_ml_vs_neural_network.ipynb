##### 1. Import Required Libraries

````python
import pandas as pd
import matplotlib.pyplot as plt
````

##### 2. Define Model Metrics

````python
logistic_regression_metrics = {
    "accuracy": 0.8041,
    "precision": 0.6644,
    "recall": 0.5294,
    "f1_score": 0.5893,
    "roc_auc": 0.8416,
}

neural_network_metrics = {
    "accuracy": 0.8032,
    "precision": 0.7039,
    "recall": 0.4484,
    "f1_score": 0.5478,
    "roc_auc": 0.8459,
}
````

##### 3. Build Comparison DataFrame

````python
comparison_df = pd.DataFrame(
    {
        "Logistic Regression":
            logistic_regression_metrics,

        "Neural Network":
            neural_network_metrics,
    }
)

comparison_df
````

##### 4. Add Metric Difference

````python
comparison_df[
    "NN - Logistic Regression"
] = (
    comparison_df["Neural Network"]
    - comparison_df["Logistic Regression"]
)

comparison_df
````

Positive difference → Neural Network higher

Negative difference → Logistic Regression higher

##### 5. Display Percentage Metrics

````python
display_df = comparison_df.copy()

display_df[
    [
        "Logistic Regression",
        "Neural Network",
        "NN - Logistic Regression",
    ]
] = (
    display_df[
        [
            "Logistic Regression",
            "Neural Network",
            "NN - Logistic Regression",
        ]
    ]
    * 100
)

display_df
````

##### 6. Visualize Performance Comparison

````python
plot_df = comparison_df[
    [
        "Logistic Regression",
        "Neural Network",
    ]
]

ax = plot_df.plot(
    kind="bar",
    figsize=(10, 6),
)

ax.set_title(
    "Logistic Regression vs Neural Network"
)

ax.set_xlabel(
    "Evaluation Metric"
)

ax.set_ylabel(
    "Score"
)

ax.set_ylim(
    0,
    1,
)

plt.xticks(
    rotation=0
)

plt.grid(
    axis="y"
)

plt.show()
````

##### 14. Compare Training Workflow

###### Logistic Regression

Much of the training process is encapsulated:

    X_train,
    y_train,
)

The library handles optimization internally.

###### Neural Network

We explicitly manage:

- Batching
- Forward propagation
- Loss
- Backpropagation
- Gradients
- Optimizer
- Epochs
- Validation
- Checkpointing
- Early stopping

Typical lifecycle:

``` text

optimizer.zero_grad()
        ↓
model(X_batch)
        ↓
loss
        ↓
loss.backward()
        ↓
optimizer.step()

```

The neural network gives us greater flexibility but requires more engineering.

##### 23. Important Fair-Comparison Limitation

Before treating the numeric differences as definitive, verify that both models were evaluated under equivalent conditions:

- Same source dataset
- Same feature definitions
- Same train/validation/test split
- Same preprocessing logic where appropriate
- Same test customers
- Comparable threshold policy
- Same metric definitions

If the Logistic Regression metrics came from an earlier experiment with a different train/test split, the comparison should be considered directional rather than strictly controlled.

A production model-selection experiment should evaluate both models on the same held-out test set.

##### Conclusion

This notebook compared the traditional Logistic Regression baseline with the PyTorch neural network for Telco Customer Churn.

The Logistic Regression model achieved approximately:

- Accuracy  = 0.8041
- Precision = 0.6644
- Recall    = 0.5294
- F1 Score  = 0.5893
- ROC-AUC   = 0.8416

The neural network achieved:

- Accuracy  = 0.8032
- Precision = 0.7039
- Recall    = 0.4484

The neural network showed slightly better ROC-AUC and stronger precision.

For this relatively small structured dataset, the neural network does not currently demonstrate a sufficiently large overall performance advantage to clearly justify replacing Logistic Regression.

The exercise nevertheless provided significant learning value by exposing the internal mechanics of neural-network training that traditional machine-learning APIs often abstract away.

The final production-model decision should be based on both predictive performance and business requirements, including the relative cost of false positives and false negatives.
