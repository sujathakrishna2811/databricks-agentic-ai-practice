##### What is a Neural Network?

A neural network is a machine-learning model made up of connected computational units called neurons.

The neurons are organized into layers.

A simple neural network may contain:

``` text

Input Layer
     ↓
Hidden Layer
     ↓
Output Layer

```

A deeper neural network may contain:

``` text

Input Layer
     ↓
Hidden Layer 1
     ↓
Hidden Layer 2
     ↓
Hidden Layer 3
     ↓
Output Layer

```

The hidden layers allow the model to learn increasingly complex patterns from the data.

##### What is a Neuron?

- A neuron is a small computational unit inside a neural network.

- A neuron performs the following steps:

``` text 

Inputs
   ↓
Weights
   ↓
Weighted Sum
   ↓
+ Bias
   ↓
Activation Function
   ↓
Output

```

Mathematically:

z = x₁w₁ + x₂w₂ + ... + xₙwₙ + b

Then:

a = f(z)

Where:

- x = input features
- w = weights
- b = bias
- z = weighted sum plus bias
- f = activation function
- a = neuron output


##### Example of a Single Neuron

Assume three customer features have already been converted into numerical values.

Inputs:

``` text

Tenure          = 0.2
MonthlyCharges  = 0.8
Contract        = 1.0

```
Assume the neuron currently has the following weights:

``` text 

Tenure          weight = -0.7
MonthlyCharges  weight = +0.5
Contract        weight = +0.9

```

Bias:

b = -0.3

These weight values are only examples used to explain the calculation.

In a real neural network, the final weights are learned during training.


##### Weights

- Each input feature is connected to a neuron through a weight.

Example:

``` text 

Tenure          0.2 × -0.7
MonthlyCharges  0.8 × +0.5
Contract        1.0 × +0.9

```

- Weights determine how strongly an input contributes to the neuron's calculation.

- Two properties are important:

    - ##### Sign

        The sign determines the direction of influence.

        - Positive weight:

                + → pushes the neuron's weighted calculation upward

        - Negative weight:

                - → pushes the neuron's weighted calculation downward

    - ##### Magnitude

        The absolute value determines the strength of the weight.

        For example:

        |0.9| > |-0.7| > |0.5|

        Therefore:

        - 0.9 has the largest magnitude.

        - -0.7 has the second largest magnitude.

        - 0.5 has the smallest magnitude.

- Important:

    - Sign = direction

    - Magnitude = strength

    - Weights are not restricted to values between 0 and 1.


##### Weighted Sum

- The neuron multiplies every input by its corresponding weight.

    - Tenure: 0.2 × -0.7 = -0.14
    - MonthlyCharges: 0.8 × 0.5 = 0.40
    - Contract: 1.0 × 0.9 = 0.90
    - Add the results: -0.14 + 0.40 + 0.90 = 1.16

Therefore: Weighted Sum = 1.16


##### Bias

A bias is another trainable parameter added to the weighted sum.

Suppose:

Bias = -0.3

Then:

z = Weighted Sum + Bias

z = 1.16 - 0.3

z = 0.86

Therefore:

z = 0.86

The bias gives the neuron additional flexibility when learning patterns from the data.

It is similar conceptually to the intercept used in regression models.


##### Activation Function

After calculating:

z = 0.86

the neuron applies an activation function.

An activation function transforms the neuron's raw calculation.

For this simple binary churn example, we use the Sigmoid activation function.


## Sigmoid Activation Function

The sigmoid function is:

σ(z) = 1 / (1 + e⁻ᶻ)

Where:

e ≈ 2.71828

The value e is a mathematical constant.

It is not learned by the neural network.

Using:

z = 0.86

we calculate:

σ(0.86) ≈ 0.703

Therefore the neuron output is approximately:

0.703

Because sigmoid produces values between 0 and 1, its output can be interpreted as a probability in binary classification.

In this example:

Churn Probability ≈ 70.3%


##### Classification Threshold

A probability alone does not give us the final class.

For binary classification we can apply a classification threshold.

A common starting threshold is:

0.5

Example:

Probability < 0.5
        ↓
No Churn

Probability >= 0.5
        ↓
Churn

Our model produced:

0.703

Since:

0.703 > 0.5

the predicted class is:

Churn

Important:

0.5 is not mandatory.

The threshold can be adjusted depending on the business problem and the trade-off between precision and recall.



##### Complete Single-Neuron Calculation

``` text

Input Features
      ↓
Multiply by Weights
      ↓
Weighted Sum
      ↓
Add Bias
      ↓
z
      ↓
Activation Function
      ↓
Probability
      ↓
Classification Threshold
      ↓
Predicted Class

```

For our example:

``` text 

Inputs
      ↓
Weighted Sum = 1.16
      ↓
Bias = -0.3
      ↓
z = 0.86
      ↓
Sigmoid
      ↓
0.703
      ↓
Threshold = 0.5
      ↓
Churn

```



##### Neural Network Layers

A neural network normally contains three types of layers:

1. Input Layer
2. Hidden Layer or Hidden Layers
3. Output Layer

Example:

``` trxt 

Input Layer
     ↓
Hidden Layer 1
     ↓
Hidden Layer 2
     ↓
Output Layer

```

The number of hidden layers and the number of neurons inside them are choices made while designing the neural network.


##### Input Layer

The input layer represents the features provided to the neural network.

Example:

``` text

Tenure
MonthlyCharges
Contract
TotalCharges

```

If the model receives four input features, the network receives four input values.

The input layer itself does not perform the same computation as hidden neurons.

It mainly represents the data entering the network.


## Hidden Layers

Hidden layers sit between the input layer and the output layer.

Example:

``` text 

4 Input Features
       ↓
Hidden Layer 1
   8 neurons
       ↓
Hidden Layer 2
   4 neurons
       ↓
Output Layer
   1 neuron

```
Hidden Layer 1 receives the original input features.

Hidden Layer 2 receives the outputs produced by Hidden Layer 1.

The final output layer receives the outputs produced by the previous hidden layer.


##### Fully Connected Layers

In a fully connected or dense neural network, every neuron in one layer receives all outputs from the previous layer.

Example:

Input Features:

``` text 

Tenure
MonthlyCharges
Contract
TotalCharges

```
Hidden Layer:

3 neurons

Each hidden neuron receives all four input features.

Therefore:

Neuron 1 receives all 4 features.

Neuron 2 receives all 4 features.

Neuron 3 receives all 4 features.

The neurons do not divide the features among themselves.


##### Why Do Different Neurons Learn Different Things?

Although neurons in the same fully connected layer receive the same inputs, each neuron has its own weights and bias.

Example:

Neuron 1:

``` text 

Tenure          × w11
MonthlyCharges  × w12
Contract        × w13
TotalCharges    × w14
                + Bias 1

```

Neuron 2:

``` text 

Tenure          × w21
MonthlyCharges  × w22
Contract        × w23
TotalCharges    × w24
                + Bias 2

```

Neuron 3:

```

Tenure          × w31
MonthlyCharges  × w32
Contract        × w33
TotalCharges    × w34
                + Bias 3
```

Because the neurons use different learned parameters, they can respond differently to the same input data.


#### Parameters vs Hyperparameters

It is important to distinguish model parameters from hyperparameters.

##### Parameters

Parameters are learned by the neural network during training.

Examples:

- Weights
- Biases

##### Hyperparameters

Hyperparameters are choices we make when designing or training the network.

Examples:

- Number of hidden layers
- Number of neurons in each hidden layer
- Learning rate
- Batch size
- Number of epochs
- Optimizer

Simple rule:

Parameters → learned by the model

Hyperparameters → chosen before or during model configuration


##### Connection Between Logistic Regression and Neural Networks

Binary logistic regression performs a calculation similar to a single output neuron using sigmoid.

The logistic regression equation can be written as:

P(y = 1) = σ(wᵀx + b)

The term:

wᵀx

is shorthand for:

w₁x₁ + w₂x₂ + ... + wₙxₙ

Therefore:

P(y = 1)

=

σ(
w₁x₁ +
w₂x₂ +
... +
wₙxₙ +
b
)

Conceptually:

``` text 

Features
   ↓
Weighted Sum
   ↓
Bias
   ↓
Sigmoid
   ↓
Probability

```

This is very similar to the single neuron studied earlier.

A neural network extends this idea by adding hidden layers containing many neurons.


##### Logistic Regression vs Neural Network

Logistic Regression:

``` text

Features
   ↓
Weighted Sum + Bias
   ↓
Sigmoid
   ↓
Prediction

```

Neural Network:

``` text

Features
   ↓
Hidden Layer 1
   ↓
Activation
   ↓
Hidden Layer 2
   ↓
Activation
   ↓
Output Layer
   ↓
Sigmoid
   ↓
Prediction

```

This makes Logistic Regression a useful bridge between traditional machine learning and neural networks.


##### How Are Weights Learned?

The final neural-network weights are not manually assigned.

Training learns them from data.

At the beginning, the network starts with initialized weights.

Example:

``` text

Tenure          → 0.12
MonthlyCharges  → -0.08
Contract        → 0.21

```

The model then performs:

``` text

Initialize Weights
       ↓
Forward Propagation
       ↓
Prediction
       ↓
Compare Prediction with Actual Value
       ↓
Calculate Loss
       ↓
Backpropagation
       ↓
Calculate Gradients
       ↓
Optimizer Updates Weights
       ↓
Repeat

```

As training continues, the network adjusts its weights and biases in an attempt to reduce prediction error.


##### Forward Propagation

Forward propagation means sending input data through the neural network from the input layer toward the output layer.

Example:

``` text 

Customer Features
       ↓
Hidden Layer 1
       ↓
Activation
       ↓
Hidden Layer 2
       ↓
Activation
       ↓
Output Layer
       ↓
Prediction

```

The result of forward propagation is the model prediction.


##### Loss

After the model makes a prediction, the prediction is compared with the actual target.

Example:

Actual Churn = 1

Predicted Probability = 0.20

The model made a poor prediction.

A loss function measures how incorrect the prediction is.

For binary classification, Binary Cross-Entropy is a commonly used loss function.

The goal of training is to reduce the loss.


##### Backpropagation

Backpropagation determines how the model parameters contributed to the prediction error.

Conceptually:

``` text 

Prediction
    ↓
Loss
    ↓
Backpropagation
    ↓
Calculate Gradients
    ↓
Determine how parameters should change

```

Backpropagation itself calculates gradients.

It does not directly perform the final weight update.


##### Optimizer

The optimizer uses the gradients calculated during backpropagation to update the model parameters.

Conceptually:

``` text 

Backpropagation
       ↓
Gradients
       ↓
Optimizer
       ↓
Update Weights and Biases

```
Examples of optimizers include:

- Gradient Descent
- Stochastic Gradient Descent
- Adam

Adam is commonly used for neural-network training.


## Epoch

An epoch means the model has processed the entire training dataset once.

Example:

``` text

Training Dataset
      ↓
Epoch 1
      ↓
Epoch 2
      ↓
Epoch 3
      ↓
...
      ↓
Epoch 20

```

During each epoch, the model continues adjusting its parameters.


##### Complete Neural Network Training Process

The complete training process can be summarized as:

``` text 

Initialize Parameters
       ↓
Forward Propagation
       ↓
Prediction
       ↓
Calculate Loss
       ↓
Backpropagation
       ↓
Calculate Gradients
       ↓
Optimizer Updates Parameters
       ↓
Repeat Across Batches
       ↓
Complete Epoch
       ↓
Repeat for More Epochs

```


##### Activation Functions in a Neural Network

Different layers may use different activation functions.

A common binary-classification architecture is:

``` text

Input Layer
      ↓
Hidden Layer 1
      ↓
ReLU
      ↓
Hidden Layer 2
      ↓
ReLU
      ↓
Output Layer
      ↓
Sigmoid
      ↓
Probability

```

Common activation functions include:

- ReLU
- Sigmoid
- Softmax

For the simple churn example in this notebook, the output neuron uses Sigmoid because churn prediction is a binary classification problem.

Activation functions will be studied in more detail later.


##### Key Terminology

###### Neuron

A computational unit that processes inputs using weights, bias, and an activation function.

###### Weight

A trainable parameter controlling how an input contributes to a neuron's calculation.

###### Bias

A trainable parameter added to the weighted sum.

###### Activation Function

A function that transforms a neuron's calculation.

###### Layer

A collection of neurons at a particular stage of the network.

###### Forward Propagation

Moving information from the inputs through the network to produce a prediction.

###### Loss

A numerical measure of prediction error.

###### Backpropagation

The process used to calculate gradients showing how parameters contributed to the loss.

###### Gradient

Information describing how a parameter should change to reduce loss.

###### Optimizer

An algorithm that uses gradients to update model parameters.

###### Epoch

One complete pass through the training dataset.


##### Key Learnings

In this notebook, we learned:

1. Deep Learning is a branch of Machine Learning based on neural networks.

2. Neural networks are composed of connected neurons organized into layers.

3. A neuron performs:

   Inputs → Weights → Bias → Activation → Output

4. Weights determine how inputs contribute to a neuron's calculation.

5. Weight sign represents direction, while absolute magnitude represents strength.

6. Bias provides additional flexibility to the neuron.

7. Activation functions transform neuron outputs.

8. Sigmoid converts a value into the range 0 to 1 and can be used for binary-classification probabilities.

9. A classification threshold converts a predicted probability into a class.

10. Fully connected neurons receive all outputs from the previous layer.

11. Different neurons can learn different patterns because they have different weights and biases.

12. Weights and biases are model parameters learned during training.

13. Number of layers, number of neurons, learning rate, batch size, and epochs are hyperparameters.

14. Forward propagation produces predictions.

15. Loss measures prediction error.

16. Backpropagation calculates gradients.

17. The optimizer uses gradients to update weights and biases.

18. An epoch represents one complete pass through the training dataset.

19. Logistic Regression provides a useful conceptual bridge to neural networks because binary Logistic Regression also uses a weighted sum, bias, and sigmoid.

%md

## Conclusion

This notebook established the conceptual foundation required to understand neural networks before implementing them in code.

The complete mental model is:

``` text 


Neuron
   ↓
Weights and Bias
   ↓
Activation
   ↓
Layers
   ↓
Forward Propagation
   ↓
Prediction
   ↓
Loss
   ↓
Backpropagation
   ↓
Gradients
   ↓
Optimizer
   ↓
Parameter Updates
   ↓
Epochs
   ↓
Training

```
