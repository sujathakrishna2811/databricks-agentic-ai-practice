###### Section 0 : Call project config notebook

###### Section 1 :  Read Source File Location

````python
volume_path = f"/Volumes/{CATALOG}/{SCHEMA}/raw_files"
display(dbutils.fs.ls(volume_path))
````

###### Section 2 : Ingest Raw CSV

````python
# Read the raw customer churn CSV file from the Unity Catalog Volume and create a Spark DataFrame for ingestion.
df = (spark.read.option("header", "true").option("inferSchema", "true").csv(f"{volume_path}/IBM_Telco_Customer_Churn.csv"))
display(df)
````

###### Section 3 : Inspect schema and records

````python
df.printSchema()

display(df.limit(10))
````

###### Section 4 : Write Bronze Delta table

````python
# Write the raw dataset into a Bronze Delta table while preserving the original source data.
df.write.format("delta").mode("overwrite").saveAsTable(BRONZE_TABLE)
````

###### Section 5 : Validate row count and schema

````python
source_count = df.count()

bronze_df = spark.table(BRONZE_TABLE)
bronze_count = bronze_df.count()

print(f"Source rows: {source_count}")
print(f"Bronze rows: {bronze_count}")

bronze_df.printSchema()

assert source_count == bronze_count
````

###### Key Learnings

- Loaded the raw CSV data into a Bronze Delta table.

- Preserved the original source data without applying any transformations.

- Validated row counts after ingestion to ensure data integrity.

###### Notebook Conclusion

- Successfully ingested the raw customer churn dataset into the Bronze Delta table while preserving the original source data. This Bronze table serves as the foundation for data cleaning and standardization in the Silver layer.
