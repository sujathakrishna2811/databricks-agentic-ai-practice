###### Section 1 : Create Schema

````python
#Creates a Unity Catalog schema to logically organize all Delta tables used in this project.

spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SCHEMA}
""")
````

###### Section 2 : Create Volume

````python

# Creates a managed Unity Catalog Volume to store raw source files before ingestion into Delta Lake.

spark.sql(f"""
CREATE VOLUME IF NOT EXISTS
{CATALOG}.{SCHEMA}.raw_files
""")
````

###### Section 3 : Validate Resources

````python
#Verifies that the schema and volume were created successfully before proceeding to data ingestion.

display(
    spark.sql(
        f"SHOW VOLUMES IN {CATALOG}.{SCHEMA}"
    )
)

display(
    spark.sql(
        f"SHOW TABLES IN {CATALOG}.{SCHEMA}"
    )
)

print("Schemas")
display(
    spark.sql(f"SHOW SCHEMAS IN {CATALOG}")
)

print("Volumes")
display(
    spark.sql(f"SHOW VOLUMES IN {CATALOG}.{SCHEMA}")
)
````
