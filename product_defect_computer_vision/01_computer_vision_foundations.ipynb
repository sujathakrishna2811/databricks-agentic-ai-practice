#### 01 - Computer Vision Foundations

Computer Vision (CV) is a field of AI that enables computers to extract useful information from images and videos.

A human sees: “This metal surface has a scratch.”

A computer initially sees: Numbers representing pixel values.

Our goal is to learn how we get from those numbers to:

``` text

Image
  ↓
Pixels
  ↓
Numerical representation
  ↓
Machine Learning / Deep Learning model
  ↓
Visual patterns learned
  ↓
Prediction

"Scratch"

```

In our project, the business problem will essentially be:

Given an image of a manufactured product/surface, predict its defect category.

##### 2. Computer Vision vs Image Processing

These two terms are related, but they're not the same thing.

**Image Processing**

Image processing means transforming or manipulating an image.

Examples include:

- Resize image
- Crop image
- Rotate image
- Adjust brightness
- Convert to grayscale
- Reduce noise
- Normalize pixels

- Input: Image
- Output: Modified image

**Computer Vision**

Computer Vision tries to understand information contained in the image.

Examples:

- Is there a defect?
- What type of defect is it?
- Where is the defect?
- What object is present?

- Input: Image
- Output: Information / prediction

A useful distinction is:

**Image Processing → change the image.**

**Computer Vision → understand the image.**


Our project uses both.

##### 3. Major Computer Vision Tasks

There are several different problems in Computer Vision.

**Image classification**

One image receives one class.

``` text

Image
  ↓
Model
  ↓
Scratch

```

This is what we'll initially build.

**Object detection**

We determine both:

``` text

What is present?

For example:

Metal surface
```

For our first project:

We are doing image classification.

Later, once classification is clear, object detection or segmentation would be natural extensions.

##### 4. What is an image to a computer?

This is one of the most important ideas in Computer Vision.

Suppose we have a tiny grayscale image: 4 × 4 pixels

A computer could represent it approximately as:

``` text

12    20    35    40
18    45    80    92
25    70   180   200
30    90   210   255

```

Every number represents the intensity of one pixel.

Typically for an 8-bit grayscale image:

- 0   = black
- 255 = white

and values between them represent different shades of gray.

So an image is fundamentally a matrix of numbers.

This should already feel familiar because neural networks don't understand concepts such as:

- "scratch"
- "metal"
- "crack"

directly.

They operate on numbers.

##### 5. What is a pixel?

**Pixel = Picture Element**

It is the smallest addressable element in a raster image.

Imagine:
``` text

Image

┌───┬───┬───┬───┐
│ P │ P │ P │ P │
├───┼───┼───┼───┤
│ P │ P │ P │ P │
├───┼───┼───┼───┤
│ P │ P │ P │ P │
└───┴───┴───┴───┘
```

Each P represents a pixel.

A photograph might contain millions of them.

For example: 1920 × 1080 contains: 2,073,600 pixels

##### 6. Color images — RGB

Color images introduce another dimension.

Most images we work with use three channels:

- R = Red
- G = Green
- B = Blue

Instead of one number describing a pixel, we now have three numbers.

For example:

Pixel

- R = 255
- G = 0
- B = 0

represents strong red.

Another pixel:

- R = 0
- G = 255
- B = 0

represents green.

And:

- R = 255
- G = 255
- B = 255

represents white in the standard 8-bit RGB representation.

Therefore an RGB image can be thought of as three matrices:

        IMAGE
``` text

   Red channel
       +
   Green channel
       +
   Blue channel

```

##### 7. Image dimensions

Suppose we resize an image to: 224 × 224

and it has RGB channels.

Conceptually we have:

- Height = 224
- Width  = 224
- Channels = 3

Many Python image libraries may represent that as: [224, 224, 3]

But PyTorch commonly represents an individual image tensor as: [3, 224, 224] or: [channels, height, width]

So:

``` text

RGB Image

224 × 224 × 3
      ↓
PyTorch Tensor
[3, 224, 224]

```

This idea will become extremely important later.

##### 9. Why don't we simply flatten the image?

Suppose our image is: 224 × 224 × 3

That contains:

224 × 224 × 3 =  150,528 values

We technically could flatten that into: [150528]

and feed it into a normal fully connected neural network.

But there is a major problem.

**Flattening largely removes the explicit spatial structure.**

In images, location matters.

Consider a scratch: ──────────────

Nearby pixels form the line.

The relationship between neighboring pixels is extremely useful.

This motivates a different type of neural network:** Convolutional Neural Network (CNN)**

CNNs are designed to exploit spatial patterns in images.

We'll learn exactly how that works rather than treating CNN as a black box.

##### 10. What does a CNN learn?

``` text

IMAGE
  ↓
Early CNN layers
  ↓
Edges / simple patterns
  ↓
Middle layers
  ↓
Textures / shapes
  ↓
Deeper layers
  ↓
More complex visual patterns
  ↓
Classifier
  ↓
Defect category

```

For our defect project, the network might learn visual signals related to:

- edges
- lines
- texture changes
- surface irregularities
- cracks
- scratches
- patterns

We don't manually program:

if long_thin_line:
    return "scratch"

Instead, the model learns useful visual features from training data through backpropagation.

That's the same fundamental learning mechanism you've already seen with neural networks.

##### 11. Our Computer Vision training pipeline


``` text

Defect images
      ↓
Load images
      ↓
Resize
      ↓
Transform / normalize
      ↓
Image tensors
      ↓
CNN
      ↓
Logits
      ↓
CrossEntropyLoss
      ↓
Backpropagation
      ↓
Optimizer updates weights
      ↓
Repeat for many epochs
      ↓
Trained defect classifier

```

Those concepts haven't changed.

We're adding a new way of **extracting features from images**.


🔍 XAI question:

What visual evidence influenced this prediction?

We aren't implementing this yet. We're simply establishing where Explainable AI fits.

##### 13. ⚖️ Responsible AI connection

Now suppose our model achieves:

Overall accuracy = 95%

Sounds excellent.

But imagine:

- Scratch recall = 98%
- Dent recall    = 96%
- Crack recall   = 55%

If cracks are especially important for product safety, **95% overall accuracy could be misleading.**

**⚖️ RAI question:**

Is this model sufficiently reliable for the intended manufacturing decision?

Responsible AI therefore isn't simply:

Is accuracy high?

It also asks questions such as:

- Where does the model fail?
- Are some categories disproportionately affected?
- What are the consequences of false negatives?
- When should a human inspect the product?
- How confident should the model be before automation?
- What limitations should users know?
- We'll explore these systematically in our dedicated RAI notebooks.

##### 14. Three areas we're learning together


``` text

COMPUTER VISION
"What does the image contain?"

        ↓

🔍 EXPLAINABLE AI
"Why did the model make that prediction?"

        ↓

⚖️ RESPONSIBLE AI
"Can we appropriately trust and use that
prediction in this situation?"

```

They are related, but they solve different problems.

##### 15. Key Learning

1. Computer Vision enables AI systems to extract useful information
   from images and videos.

2. Image processing transforms images, while Computer Vision attempts
   to understand their contents.

3. Image classification assigns a class to an image, object detection
   identifies objects and their locations, and segmentation identifies
   relevant regions at the pixel level.

4. Digital images are numerical data composed of pixels.

5. A grayscale pixel can be represented by an intensity value, commonly
   from 0 to 255 for an 8-bit image.

6. RGB images contain Red, Green, and Blue channels.

7. A PyTorch image tensor commonly uses the structure:
   [channels, height, width].

8. Images contain spatial relationships between neighboring pixels.
   CNNs are designed to learn from these spatial patterns.

9. CNNs can learn increasingly complex visual features, progressing
   from simple patterns such as edges toward higher-level visual
   representations.

10. Neural-network concepts previously learned — forward propagation,
    loss, backpropagation, optimizers, epochs and logits — also apply
    to Computer Vision models.

11. 🔍 Explainable AI helps investigate why a vision model made a
    particular prediction.

12. ⚖️ Responsible AI helps evaluate whether the model is sufficiently
    reliable, safe, transparent and appropriate for its intended use.

##### Conclusion

**To us, an image contains objects, textures and defects. To the computer, it begins as numerical pixel values. Computer Vision models learn patterns in those numbers that allow them to recognize meaningful visual features.**
